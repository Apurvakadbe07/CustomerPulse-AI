```python
import pandas as pd

customers = pd.read_csv(
    path + "olist_customers_dataset.csv"
)

orders = pd.read_csv(
    path + "olist_orders_dataset.csv"
)

order_items = pd.read_csv(
    path + "olist_order_items_dataset.csv"
)

payments = pd.read_csv(
    path + "olist_order_payments_dataset.csv"
)

reviews = pd.read_csv(
    path + "olist_order_reviews_dataset.csv"
)

products = pd.read_csv(
    path + "olist_products_dataset.csv"
)

sellers = pd.read_csv(
    path + "olist_sellers_dataset.csv"
)

```

## Key Observations

### Missing Values

During the initial data understanding phase, missing values were identified in the following datasets:

- **Orders Dataset:** Missing delivery and approval timestamps indicate orders that may have been cancelled, unavailable, or not yet delivered.
- **Reviews Dataset:** Missing review titles and comments suggest that many customers submitted only a rating without textual feedback.
- **Products Dataset:** Some product attributes are missing, indicating incomplete product information.

At this stage, missing values are only identified and documented. They will be handled appropriately during the **Data Cleaning & Preprocessing** phase.

```python
# ==========================================================
# Missing Values Count
# ==========================================================

missing_count = []

for name, df in datasets.items():

    for column in df.columns:

        missing = df[column].isnull().sum()

        if missing > 0:

            missing_count.append({

                "Dataset": name,

                "Column": column,

                "Missing Values": missing

            })

missing_count = pd.DataFrame(missing_count)

missing_count.sort_values(
    by="Missing Values",
    ascending=False
).reset_index(drop=True)
```

```python
# ==========================================================
# Missing Values Percentage
# ==========================================================

missing_percentage = []

for name, df in datasets.items():

    for column in df.columns:

        missing = df[column].isnull().sum()

        if missing > 0:

            missing_percentage.append({

                "Dataset": name,

                "Column": column,

                "Missing %": round(
                    (missing / len(df)) * 100,
                    2
                )

            })

missing_percentage = pd.DataFrame(missing_percentage)

missing_percentage.sort_values(
    by="Missing %",
    ascending=False
).reset_index(drop=True)
```

```python
# ==========================================================
# Duplicate Records
# ==========================================================

duplicate_summary = []

for name, df in datasets.items():

    duplicate_summary.append({

        "Dataset": name,

        "Duplicate Rows": df.duplicated().sum()

    })

duplicate_summary = pd.DataFrame(duplicate_summary)

duplicate_summary
```
