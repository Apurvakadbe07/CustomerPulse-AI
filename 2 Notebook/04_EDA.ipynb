```python
#Cell 4 – Skewness Table
# Calculate skewness

skewness = pd.DataFrame({
    "Skewness": df[numerical_columns].skew()
})

skewness = skewness.sort_values(
    by="Skewness",
    ascending=False
)

display(skewness)
```

```python
#Cell 5 – Distribution Classification
# Function to classify skewness

def classify_skewness(value):

    if value > 1:
        return "Highly Right Skewed"

    elif value > 0.5:
        return "Moderately Right Skewed"

    elif value < -1:
        return "Highly Left Skewed"

    elif value < -0.5:
        return "Moderately Left Skewed"

    else:
        return "Approximately Symmetric"


skewness["Distribution"] = (
    skewness["Skewness"]
    .apply(classify_skewness)
)

display(skewness)
```

## Skewness Analysis

Skewness measures the asymmetry of a feature's distribution.

### Interpretation

- **Skewness > 1** → Highly Right Skewed
- **0.5 to 1** → Moderately Right Skewed
- **-0.5 to 0.5** → Approximately Symmetric
- **-1 to -0.5** → Moderately Left Skewed
- **< -1** → Highly Left Skewed

### Key Observations

- Binary indicator features (0/1) such as `repeat_customer`, `vip_customer`, and `loyal_customer` also exhibit high skewness. This is expected due to class imbalance and does not necessarily indicate a data quality issue.
- Features such as `average_review_score` and `on_time_delivery_rate` are left-skewed because most customers have high review scores and on-time deliveries.

# Part 4 – Outlier Analysis

## Objective

This section identifies potential outliers in important numerical features using box plots and the Interquartile Range (IQR) method.

Outliers are not removed automatically. Each feature is evaluated based on business logic to determine whether the extreme values represent genuine customer behaviour or data quality issues.

This ensures that valuable business information is preserved while preparing the dataset for machine learning.

```python
#Cell 1 – Select Features for Outlier Analysis
outlier_features = [
    "total_spent",
    "average_order_value",
    "maximum_order_value",
    "minimum_order_value",
    "average_payment_installments",
    "average_delivery_days",
    "average_delivery_delay",
    "recency_days",
    "customer_tenure_days",
    "average_purchase_gap",
    "spending_intensity"
]

print(f"Total Features for Outlier Analysis: {len(outlier_features)}")
outlier_features
```

```python
#Cell 2 – Box Plots
import math

n_cols = 2
n_rows = math.ceil(len(outlier_features) / n_cols)

fig, axes = plt.subplots(n_rows, n_cols, figsize=(14, n_rows * 4))
axes = axes.flatten()

for i, column in enumerate(outlier_features):

    axes[i].boxplot(df[column], vert=True)

    axes[i].set_title(column.replace("_", " ").title())

for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.tight_layout()
plt.show()
```

```python
# Calculate outliers using IQR

outlier_summary = []

for column in outlier_features:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - (1.5 * IQR)
    upper_bound = Q3 + (1.5 * IQR)

    outlier_count = (
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ).sum()

    outlier_summary.append({
        "Feature": column,
        "Q1": round(Q1, 2),
        "Q3": round(Q3, 2),
        "Lower Bound": round(lower_bound, 2),
        "Upper Bound": round(upper_bound, 2),
        "Outlier Count": outlier_count,
        "Outlier %": round(outlier_count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
#Cell 4 – Sort by Highest Outlier Percentage
outlier_summary = outlier_summary.sort_values(
    by="Outlier %",
    ascending=False
)

display(outlier_summary)
```

```python
#Cell 5 – Outlier Percentage Visualization
plt.figure(figsize=(10,6))

plt.barh(
    outlier_summary["Feature"],
    outlier_summary["Outlier %"]
)

plt.xlabel("Outlier Percentage")
plt.ylabel("Features")
plt.title("Outlier Percentage by Feature")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
#Cell 6 – Business Decision Table
decision_table = pd.DataFrame({

    "Feature": outlier_features,

    "Decision": [
        "Keep",
        "Keep",
        "Keep",
        "Keep",
        "Keep",
        "Investigate",
        "Investigate",
        "Keep",
        "Keep",
        "Keep",
        "Keep"
    ],

    "Reason": [

        "High-value customers are genuine.",

        "Premium orders are expected.",

        "Business-valid expensive purchases.",

        "No data quality issue.",

        "Higher EMI installments are valid.",

        "Possible operational delays.",
        "Contains both early and late deliveries.",

        "Old customers naturally have higher recency.",

        "Long-term customers are valuable.",

        "Large purchase gaps are normal.",

        "High spending intensity represents premium customers."
    ]
})

decision_table
```

## Outlier Analysis Summary

### Key Findings

- Spending-related variables contain approximately 8% statistical outliers. These correspond to genuine high-value customers and should be retained.
- Delivery-related variables exhibit moderate outlier percentages, reflecting operational delays and early deliveries. These features require further investigation before model training.
- Customer tenure and purchase gap show outliers because the majority of customers are one-time buyers, resulting in an Interquartile Range (IQR) of zero.
- Recency Days contains almost no extreme values and does not require any treatment.

# Part 5 – Correlation Analysis

## Objective

The objective of this section is to understand the relationships between numerical features using Pearson Correlation.

This analysis helps to:

- Identify relationships between variables
- Detect multicollinearity
- Support feature selection
- Understand customer behaviour patterns

Correlation values range from **-1 to +1**:

- **+1** → Strong Positive Correlation
- **0** → No Correlation
- **-1** → Strong Negative Correlation

> Correlation helps identify related features but does not imply causation.

```python
#Cell 1 – Import Library
import seaborn as sns
```

```python
#Cell 2 – Select Continuous Numerical Features
continuous_features = [
    "total_orders",
    "delivered_orders",
    "cancelled_orders",
    "total_spent",
    "average_order_value",
    "maximum_order_value",
    "minimum_order_value",
    "average_payment_installments",
    "total_items_purchased",
    "unique_products",
    "unique_categories",
    "average_items_per_order",
    "average_review_score",
    "average_delivery_days",
    "average_delivery_delay",
    "recency_days",
    "customer_tenure_days",
    "purchase_frequency",
    "average_purchase_gap",
    "spending_intensity"
]

print(f"Total Continuous Features: {len(continuous_features)}")
```

```python
#Cell 3 – Correlation Matrix
correlation_matrix = df[continuous_features].corr()

display(correlation_matrix.round(2))
```

```python
# Create upper triangle mask
mask = np.triu(np.ones_like(correlation_matrix, dtype=bool))

plt.figure(figsize=(14,10))

sns.heatmap(
    correlation_matrix,
    mask=mask,
    cmap="coolwarm",
    center=0,
    linewidths=0.5,
    square=True,
    annot=False,
    cbar_kws={"label": "Correlation Coefficient"}
)

plt.title(
    "Correlation Heatmap of Continuous Features",
    fontsize=16,
    fontweight="bold"
)

plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
#Cell 5 – Highly Correlated Feature Pairs
corr_pairs = correlation_matrix.unstack().reset_index()

corr_pairs.columns = ["Feature 1", "Feature 2", "Correlation"]

# Remove self correlation
corr_pairs = corr_pairs[
    corr_pairs["Feature 1"] != corr_pairs["Feature 2"]
]

# Remove duplicate pairs
corr_pairs["Pair"] = corr_pairs.apply(
    lambda x: tuple(sorted([x["Feature 1"], x["Feature 2"]])),
    axis=1
)

corr_pairs = corr_pairs.drop_duplicates("Pair")

# Keep only strong correlations
high_corr = corr_pairs[
    corr_pairs["Correlation"].abs() >= 0.80
].sort_values(
    by="Correlation",
    ascending=False
)

high_corr = high_corr.drop(columns="Pair")

display(high_corr)
```

## Business Insights

### Key Findings

- Spending-related features exhibit extremely strong positive correlations (>0.95), indicating that they capture similar aspects of customer purchasing behaviour.
- Customer Tenure and Average Purchase Gap show a very high positive correlation (0.985), reflecting long-term customer engagement patterns.
- Total Items Purchased and Average Items Per Order are also strongly correlated, suggesting similar purchasing behaviour.

### Business Decision

- No features will be removed during EDA.
- Highly correlated features will be reviewed during the feature selection stage before machine learning.
- Feature importance and model performance will guide the final decision on retaining or removing correlated variables.

## Conclusion

Correlation analysis revealed several highly correlated feature pairs, particularly among spending-related variables.

These relationships are expected in an e-commerce customer dataset and provide valuable business insights.

At this stage, no features are removed. Feature selection will be performed before model training to reduce multicollinearity while preserving business information.

```python
#Cell 6 – Top Correlated Feature Pairs
plt.figure(figsize=(10,6))

plt.barh(
    high_corr["Feature 1"] + " ↔ " + high_corr["Feature 2"],
    high_corr["Correlation"]
)

plt.xlabel("Correlation Coefficient")
plt.ylabel("Feature Pairs")
plt.title("Highly Correlated Feature Pairs")

plt.gca().invert_yaxis()

plt.tight_layout()

plt.show()
```

## Business Insights

### Key Observations

- Spending-related variables are expected to show strong positive correlations because they measure similar purchasing behaviour.
- Order-related variables are also likely to be strongly correlated.
- Strong positive correlations indicate potential feature redundancy.
- Negative correlations indicate inverse relationships between customer behaviour metrics.

### Business Decision

No features will be removed during EDA.

Highly correlated features will be reviewed before machine learning to avoid multicollinearity and improve model performance.

## Conclusion

Correlation analysis provides valuable insights into the relationships between customer behaviour variables.

The identified correlations will support feature selection and help reduce redundancy during the machine learning phase.

No feature engineering or feature removal is performed at this stage.

```python
# Cell 11 — Segment-wise Spending and Order Comparison

segment_comparison = (
    customer_360
    .groupby("spending_segment", observed=False)
    [
        [
            "total_spent",
            "total_orders"
        ]
    ]
    .mean()
    .reset_index()
)


segment_comparison_melted = segment_comparison.melt(
    id_vars="spending_segment",
    value_vars=[
        "total_spent",
        "total_orders"
    ],
    var_name="Metric",
    value_name="Average Value"
)


plt.figure(figsize=(10,5))

sns.barplot(
    data=segment_comparison_melted,
    x="spending_segment",
    y="Average Value",
    hue="Metric"
)


plt.title(
    "Purchase Behaviour Comparison Across Customer Segments",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Customer Segment")
plt.ylabel("Average Value")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

```python
# Cell 12 — Segment Distribution Percentage

segment_count = (
    customer_360["spending_segment"]
    .value_counts()
    .reset_index()
)

segment_count.columns = [
    "Customer Segment",
    "Customer Count"
]


segment_count["Percentage"] = (
    segment_count["Customer Count"] /
    segment_count["Customer Count"].sum()
) * 100


segment_count
```

```python
# Cell 13 — Customer Count Across Segments

plt.figure(figsize=(8,5))

sns.barplot(
    data=segment_count,
    x="Customer Segment",
    y="Customer Count"
)

plt.title(
    "Customer Distribution Across Spending Segments",
    fontsize=14,
    fontweight="bold"
)

plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.tight_layout()

plt.show()
```

```python
# Cell 14 — Final Segment Profile Summary

final_segment_profile = (
    customer_360
    .groupby("spending_segment", observed=False)
    .agg(
        customers=("customer_unique_id", "count"),
        avg_orders=("total_orders", "mean"),
        avg_spending=("total_spent", "mean"),
        avg_order_value=("average_order_value", "mean"),
        avg_review_score=("average_review_score", "mean")
    )
    .reset_index()
)


final_segment_profile
```
